```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import pandas as pd
import torch
```

```python
with open("names.txt","r") as f :
    words=f.read().splitlines()
f.close()
```

```python
len(words)
```

```python
min(len(w) for w in words)
```

```python
max(len(w) for w in words)
```

```python
chars=sorted(list(set(''.join(words))))
stoi={s:i for i,s in enumerate(chars,start=1)}
stoi['.']=0
itos={i:s for s,i in stoi.items() }
```

```python
N=torch.zeros((27,27),dtype=torch.int32)
for w in words :
    chs=['.']+list(w)+['.']
    for ch1,ch2 in zip(chs,chs[1:]) :
        i=stoi[ch1]
        j=stoi[ch2]
        N[i,j]+=1
```

```python
P=N.float()
P/=P.sum(dim=1,keepdim=True)
```

```python
g=torch.Generator().manual_seed(2147483647)
for _ in range(20) :
    hh=""
    ix=0
    while True :
        p=P[ix]
        p=p/p.sum()
        ix=torch.multinomial(p,num_samples=1,replacement=True,generator=g).item()
        hh+=itos[ix]
        if ix==0 :
            break
    print(hh)
```

```python
xs=[]
ys=[]
for w in words :
    chs=['.']+list(w)+['.']
    for ch1,ch2 in zip(chs,chs[1:]) :
        i=stoi[ch1]
        j=stoi[ch2]
        xs.append(i)
        ys.append(j)
xs=torch.tensor(xs)
ys=torch.tensor(ys)
```

```python
import torch.nn.functional as F
xenc=F.one_hot(xs,num_classes=27).float()
xenc.shape
```

```python
#initialization
W=torch.randn((27,27),requires_grad=True)
```

```python
#forward pass
logits=xenc @ W
counts=logits.exp()
probs=counts/counts.sum(dim=1,keepdim=True)
loss=-probs[torch.arange(228146),ys].log().mean()
loss.data
```

```python
#backward pass
W.grad=None
loss.backward()
```

```python
#update
W.data+=-0.1*W.grad
```

```python
for k in range(100) :
    #forward pass
    logits=xenc @ W
    counts=logits.exp()
    probs=counts/counts.sum(dim=1,keepdim=True)
    loss=-probs[torch.arange(228146),ys].log().mean()
    #backward pass
    W.grad=None
    loss.backward()
    #update
    W.data+=-50*W.grad
    print(k,loss.item())
```

```python
g=torch.Generator().manual_seed(2147483647)
for _ in range(20) :
    hh=""
    ix=0
    while True :
        xenc=F.one_hot(torch.tensor([ix]),num_classes=27).float()
        logits=xenc @ W
        counts=logits.exp()
        probs=counts/counts.sum(dim=1,keepdim=True)
        ix=torch.multinomial(probs,num_samples=1,replacement=True,generator=g).item()
        hh+=itos[ix]
        if ix==0 :
            break
    print(hh)
```
